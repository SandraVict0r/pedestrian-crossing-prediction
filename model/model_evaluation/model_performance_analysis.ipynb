

# 1.Importations & Fonctions utilitaires

Cette cellule initialise l’environnement de travail et définit les fonctions génériques nécessaires au chargement et à la préparation des données comportementales issues de l'expérience VR (Expérience 2).

## **Contenu de la cellule**

* **Importations** :
  Bibliothèques scientifiques utilisées pour la modélisation (NumPy, pandas, scikit-learn), la visualisation (Seaborn, Matplotlib) et les analyses statistiques (SciPy).
* **Configuration graphique** :
  Style uniforme (Times New Roman, layout IEEE-like) pour assurer une cohérence visuelle dans les figures du rapport de thèse.
* **`load_csv_files()`** :

  * Parcourt le dossier contenant les 9 CSV processed.
  * Déduit la météo à partir du nom du fichier.
  * Concatène tous les fichiers dans un unique DataFrame.
  * Ajoute les colonnes dérivées :

    * `velocity_ms` (conversion km/h → m/s)
    * `avg_safety_time` = distance / vitesse
* **`prepare_data()`** :

  * Filtre les valeurs extrêmes du seuil comportemental (`avg_safety_time`).
  * Génère les features quadratiques ou interactionnelles utilisés dans le modèle analytique :

    * `height^2`
    * `velocity_exp2^2`
    * `height*velocity`

Ces deux fonctions structurent entièrement la pipeline de préparation des données avant la modélisation du seuil comportemental minimal ($T_{end}$).


````python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from scipy.stats import pearsonr, shapiro

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "text.usetex": True,
    "axes.labelsize": 12,
    "axes.titlesize": 14,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.dpi": 300,
    "savefig.dpi": 300,
    "figure.autolayout": True
})

def load_csv_files(folder_path):
    all_files = [os.path.join(folder_path, file) for file in os.listdir(folder_path) if file.endswith('.csv')]
    datas = pd.DataFrame()
    for file in all_files:
        filename = os.path.basename(file)
        if "_" in filename:
            weather = filename.split('_')[0]
            data = pd.read_csv(file)
            data['weather'] = weather
            datas = pd.concat([datas, data], ignore_index=True)
    datas['velocity_ms'] = datas['velocity_exp2'] * (5 / 18)
    datas['avg_safety_time'] = datas['avg_safety_distance'] / datas['velocity_ms']
    return datas.dropna()

def prepare_data(datas, limit=None):
    if limit is not None:
        datas = datas[datas['avg_safety_time'] < limit]
    datas['height^2'] = datas['height'] ** 2
    datas['velocity_exp2^2'] = datas['velocity_exp2'] ** 2
    datas['height*velocity'] = datas['height'] * datas['velocity_exp2']
    return datas
````

# Entraînement du modèle + Split + Cross-Validation

Cette cellule implémente l’ensemble du pipeline scientifique utilisé pour estimer le seuil comportemental minimal
$$T_{end}$$
à partir des données VR. Elle correspond à la version expérimentale du pipeline présent dans `model_training/train.py`.

###  1. Définition des variables

* **Features analytiques** :
  $
  \text{height}, \quad \text{height}^2, \quad \text{velocity}
  $
* **Target** :
  $
  T_{end} = \text{avg safety time}
  $
* Colonnes auxiliaires : météo (`weather`) et participant (`participant_id`).


###  2. Split Train/Test (80/20)

Le jeu de données est séparé au niveau **participant**, afin d’éviter toute fuite d’information entre apprentissage et test.


###  3. Validation croisée (KFold = 5)

Pour chaque fold :

1. Un modèle linéaire est entraîné :
   $
   T_{pred} = a h + b h^2 + c v + \text{intercept}
   $

2. Les coefficients sont enregistrés pour estimer leur variance.

3. Pour chaque météo, un coefficient multiplicatif
   $
   \alpha_w = \frac{\mathbb{E}[T_{end}]}{\mathbb{E}[\hat{T}_{\text{pred}}]}
   $
   est estimé.
   Ce facteur représente la **correction perceptive météo**.

On obtient ainsi :

* la moyenne ± écart-type de ($a, b, c, \text{intercept}$),
* la moyenne ± écart-type des ($\alpha_w$).


###  4. Entraînement final sur les 80% du train

Un modèle linéaire unique est réentraîné, puis les coefficients météo finaux ($\alpha_w$) sont recalculés.

L'équation du modèle devient, pour chaque météo (w) :

$
T_{end}^{(w)} = \alpha_w \left( a h + b h^2 + c v + \text{intercept} \right)
$


###  5. Évaluation sur le test set (20%)

Les prédictions brutes sont corrigées par la météo :

$
T_{end}^{\text{pred}} = \alpha_w \cdot \hat{T}_{\mathrm{pred}}
$

Les métriques rapportées :

* **MAE**
* **RMSE**
* **R²**


###  6. Analyse des erreurs par météo

Pour chaque condition (clear, rain, night), la cellule calcule :

* l’écart-type de l’erreur ($\sigma_w$),
* la moyenne de l’erreur ($\mu_w$),

nécessaires à l'ajustement comportemental final :

$
T_{end}^{\text{adj}} = T_{end}^{\text{pred}} - 2\sigma_w + \mu_w
$

###  7. Construction de `df_bin`

Ce DataFrame regroupe :

* ($T_{end}^{\text{true}}$)
* ($T_{end}^{\text{pred}}$)
* ($T_{end}^{\text{adj}}$)
* participant, hauteur, vitesse, météo

Il sert à construire les graphes d’analyse de performance.

###  8. Appel de la fonction + affichage

La cellule :

```python
txt, X_test, y_test, weather_test, y_test_pred, participant_test, df_bin, df_train = modeling_and_plots(datas_prepared)
print(txt)
```

exécute tout le pipeline et affiche un résumé clair :

* coefficients du modèle,
* coefficients météo,
* équation factorisée finale,
* performances sur le test set.


````python
def modeling_and_plots(datas,has_mask=False):

    features = ['height', 'height^2', 'velocity_exp2']
    target = 'avg_safety_time'
    weather_col = 'weather'
    participant_col = 'participant_id'


    X = datas[features]
    y = datas[target]
    weather = datas[weather_col]

    # Split 80/20
    participants = datas[participant_col]

    X_train, X_test, y_train, y_test, weather_train, weather_test, participant_train, participant_test = train_test_split(
        X, y, weather, participants, test_size=0.2, random_state=42
    )


    # Cross-validation 10 folds
    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    alphas_cv = {w: [] for w in weather.unique()}
    coefs_cv = []

    for train_idx, val_idx in kf.split(X_train):
        X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
        y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
        weather_val = weather_train.iloc[val_idx]

        model = LinearRegression()
        model.fit(X_tr, y_tr)

        y_val_pred = model.predict(X_val)
        coefs_cv.append(np.append(model.coef_, model.intercept_))

        df_val = pd.DataFrame({'y_true': y_val, 'y_pred': y_val_pred, 'weather': weather_val})
        for w in alphas_cv.keys():
            subset = df_val[df_val['weather'] == w]
            if len(subset) > 0:
                alpha = subset['y_true'].mean() / subset['y_pred'].mean()
                alphas_cv[w].append(alpha)

    coefs_cv = np.array(coefs_cv)
    coef_means = coefs_cv[:, :3].mean(axis=0)
    coef_stds = coefs_cv[:, :3].std(axis=0)
    intercept_mean = coefs_cv[:, 3].mean()
    intercept_std = coefs_cv[:, 3].std()

    alpha_means = {w: np.mean(vals) for w, vals in alphas_cv.items()}
    alpha_stds = {w: np.std(vals) for w, vals in alphas_cv.items()}

    # Texte résultats
    txt = "=== Résultats Cross-validation ===\n"
    txt += "Coefficients fixes (mean ± std):\n"
    for var, mean_, std_ in zip(features, coef_means, coef_stds):
        txt += f"  {var}: {mean_:.4f} ± {std_:.4f}\n"
    txt += f"  intercept: {intercept_mean:.4f} ± {intercept_std:.4f}\n\n"
    txt += "Coefficients multiplicateurs alpha par météo (mean ± std):\n"
    for w in alpha_means.keys():
        txt += f"  {w}: {alpha_means[w]:.4f} ± {alpha_stds[w]:.4f}\n"

    # Entraînement final sur train complet
    final_model = LinearRegression()
    final_model.fit(X_train, y_train)

    y_train_pred = final_model.predict(X_train)
    df_train = pd.DataFrame({'y_true': y_train, 'y_pred': y_train_pred, 'weather': weather_train})

    alphas_final = {}
    for w in alpha_means.keys():
        subset = df_train[df_train['weather'] == w]
        alphas_final[w] = subset['y_true'].mean() / subset['y_pred'].mean()

    txt += "\n=== Équation factorisée finale après entraînement complet sur train 80% ===\n"
    coefs = final_model.coef_
    intercept = final_model.intercept_
    for w, alpha in alphas_final.items():
        eq_terms = " + ".join([f"{c:.4f}*{f}" for c, f in zip(coefs, features)])
        txt += f"{w} : y = {alpha:.4f} * ({eq_terms} + {intercept:.4f})\n"


    # Évaluation sur test
    y_test_pred_raw = final_model.predict(X_test)
    alpha_array = weather_test.map(alphas_final).values
    y_test_pred = pd.Series(y_test_pred_raw * alpha_array, index=y_test.index)

    # Filtrer les valeurs où y_test <= 7
    if has_mask :
        mask = y_test <= 5
        y_test_filtered = y_test[mask]
        y_pred_filtered = y_test_pred[mask]

        mae_test = mean_absolute_error(y_test_filtered, y_pred_filtered)
        rmse_test = mean_squared_error(y_test_filtered, y_pred_filtered)
        r2_test = r2_score(y_test_filtered, y_pred_filtered)
    else :
        mae_test = mean_absolute_error(y_test, y_test_pred)
        rmse_test = mean_squared_error(y_test, y_test_pred)
        r2_test = r2_score(y_test, y_test_pred)


    txt += "\n=== Évaluation sur test 20% ===\n"
    txt += f"MAE: {mae_test:.4f}\n"
    txt += f"RMSE: {rmse_test:.4f}\n"
    txt += f"R²: {r2_test:.4f}\n"

    # --- Calcul std et mean erreur par météo ---
    df_error = pd.DataFrame({'weather': weather_test, 'error': y_test - y_test_pred})
    df_error = df_error[y_test <= 7]
    
    # Pour chaque météo, calcul std et mean des erreurs
    std_me_per_weather = {}
    for w in df_error['weather'].unique():
        subset_err = df_error[df_error['weather'] == w]['error']
        std_me_per_weather[w] = {
            'std': subset_err.std(),
            'me': subset_err.mean()
        }
        
        
    # --- Binned trend plot: test set points + baseline + 5-bin average with standard error ---
    df_bin = pd.DataFrame({
        'y_true': y_test,
        'y_pred': y_test_pred
    })

    df_bin['participant'] = participant_test.values
    df_bin['height'] = X_test['height'].values
    df_bin['velocity_exp2'] = X_test['velocity_exp2'].values
    df_bin['weather'] = weather_test.values


    # Ajout des prédictions ajustées par météo (pred - 2*std + me)
    y_pred_adjusted = y_test_pred.copy()
    for w in std_me_per_weather.keys():
        idxs = weather_test[weather_test == w].index
        std_ = std_me_per_weather[w]['std']
        me_ = std_me_per_weather[w]['me']
        y_pred_adjusted.loc[idxs] = y_pred_adjusted.loc[idxs] - 2 * std_ + me_

    df_bin['y_pred_adjusted'] = y_pred_adjusted

        
    return txt, X_test, y_test, weather_test, y_test_pred, participant_test, df_bin, df_train

````

````python
folder_path = r'C:\Users\svictor\Documents\model_datas'
datas = load_csv_files(folder_path)
datas_prepared = prepare_data(datas, 7)
txt, X_test, y_test, weather_test, y_test_pred, participant_test, df_bin, df_train = modeling_and_plots(datas_prepared, has_mask=False)
print(txt)
````

#  Analyse des résidus + Graphes

Cette cellule examine la qualité de la prédiction du seuil comportemental minimal
$$T_{end}$$
en analysant les résidus du test set :

$
\text{résidu} = y_{\text{true}} - y_{\text{pred}}
$

L’objectif est de vérifier :

* la stabilité du modèle,
* l’absence de biais systématique,
* la distribution des erreurs,
* le caractère gaussien des résidus (critère important pour un modèle linéaire).

Le premier graphique affiche :

* l’histogramme des résidus sur le test set ;
* la courbe KDE (lissage) ;
* une ligne verticale rouge pointillée représentant un résidu nul ;
* une ligne bleue représentant la moyenne des résidus.

Une moyenne proche de zéro indique que le modèle n’est pas globalement biaisé dans une direction (surestimation ou sous-estimation de $T_{end}$).

La cellule calcule ensuite :

* **MAE** :erreur absolue moyenne
* **Corrélation de Pearson** entre valeurs réelles et prédites
* **Test de Shapiro–Wilk** sur les résidus, pour vérifier leur normalité

$
H_0 : \text{les résidus suivent une loi normale}
$

Un p-value > 0.05 indique que l’hypothèse de normalité **n’est pas rejetée**, ce qui valide l’hypothèse d’un bruit additif gaussien dans le modèle.


* Une moyenne des résidus ≈ 0 → pas de biais global
* Une distribution centrée mais légèrement asymétrique → erreurs hétérogènes selon participants/vitesse/météo
* Normalité plausible → le modèle linéaire est cohérent avec la structure des données



````python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import shapiro, pearsonr
from sklearn.metrics import mean_absolute_error

# IEEE-like style with sober, visible colors
sns.set_theme(style="whitegrid", palette=['navy', 'darkred', 'dimgray'])

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "text.usetex": True,
    "axes.labelsize": 12,
    "axes.titlesize": 14,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "figure.dpi": 300,
    "figure.autolayout": True
})

# --- Residuals ---
residuals = y_test - y_test_pred

plt.figure(figsize=(6, 4))
sns.histplot(residuals, bins=30, kde=True, color='dimgray', edgecolor='white', linewidth=0.5)
plt.axvline(0, color='darkred', linestyle='--', label='Zero Residual')
plt.axvline(residuals.mean(), color='navy', linestyle='-', label=fr'Mean = {residuals.mean():.3f}')
plt.title(r'Residuals Distribution (Test Set)')
plt.xlabel(r'Residual ($y_\mathrm{true} - y_\mathrm{pred}$)')
plt.ylabel(r'Frequency')
plt.legend()
plt.grid(True)
plt.show()

# --- Statistics ---
mae = mean_absolute_error(y_test, y_test_pred)
r, pval = pearsonr(y_test, y_test_pred)
stat, p_shapiro = shapiro(residuals)

print(f"MAE = {mae:.3f}")
print(f"Pearson correlation r = {r:.3f}, p-value = {pval:.3e}")
print(f"Shapiro-Wilk stat = {stat:.3f}, p = {p_shapiro:.3e}")

if p_shapiro > 0.05:
    print("✅ Residuals appear normally distributed (p > 0.05)")
else:
    print("⚠️ Residuals do not appear normal (p ≤ 0.05)")
````

## Prédictions vs valeurs réelles (Test Set)

Cette cellule visualise la capacité du modèle à prédire correctement le seuil comportemental minimal
$$T_{end}$$
sur le jeu de test.

L’objectif : vérifier que les prédictions suivent bien la diagonale idéale
$
y_{\text{pred}} = y_{\text{true}}
$
et analyser la stabilité du modèle via un *binned trend*.

###  1. Scatter plot : structure des prédictions

Chaque point bleu représente :

* une observation du test set,
* avec en abscisse le (T_{end}) réel,
* et en ordonnée le (T_{end}) prédit.

Une bonne cohérence entre modèle et données se manifeste par un nuage réparti autour de la diagonale rouge :

$
y = x
$

###  2. Binned trend (5 intervalles)

Pour réduire le bruit et visualiser la tendance globale du modèle, les données sont regroupées en 5 bins selon ($T_{end}$) réel.

Pour chaque bin, la figure affiche :

* le **$T_{end}$ moyen réel**,
* le **$T_{end}$ moyen prédit**,
* une **barre d’erreur** correspondant à l’erreur standard
  $
  \text{SE} = \frac{\sigma}{\sqrt{n}}
  $

Cette courbe orange permet d’évaluer :

* la stabilité du modèle,
* le respect de la monotonie (plus le seuil réel augmente, plus la prédiction doit augmenter),
* la présence éventuelle de plafonnement ou de déviation systématique.


###  3. Interprétation générale

Dans ce type de graphique, on cherche à vérifier que :

* la courbe binned suit ou épouse la diagonale (y = x),
* l’écart moyen reste faible,
* les barres d’erreur ne deviennent pas disproportionnées pour un bin particulier.


````python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd

# Assume df_bin contient 'y_true' et 'y_pred'

nb_bins = 5
df_bin['bin'] = pd.cut(df_bin['y_true'], bins=nb_bins)

bin_stats = df_bin.groupby('bin').agg(
    mean_y_true=('y_true', 'mean'),
    mean_y_pred=('y_pred', 'mean'),
    std_y_pred=('y_pred', 'std'),
    count=('y_pred', 'count')
).reset_index()

bin_stats['std_error'] = bin_stats['std_y_pred'] / np.sqrt(bin_stats['count'])

min_val = min(df_bin['y_true'].min(), df_bin['y_pred'].min())
max_val = max(df_bin['y_true'].max(), df_bin['y_pred'].max())

plt.figure(figsize=(6,6))

palette = sns.color_palette("deep")

scatter_color = palette[0]   # Couleur 1 pour scatter points
binned_color = palette[1]    # Couleur 2 pour binned trend + barres erreur (même couleur)

plt.scatter(df_bin['y_true'], df_bin['y_pred'], alpha=0.6, color=scatter_color, label='Test set')

plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=1.5, label=r'$y = x$')

plt.errorbar(
    bin_stats['mean_y_true'],
    bin_stats['mean_y_pred'],
    yerr=bin_stats['std_error'],
    fmt='o-', color=binned_color, ecolor=binned_color,
    elinewidth=2, capsize=4,
    label='Model Prediction'
)

plt.title(r'Predicted vs Actual Values (Test Set)')
plt.xlabel(r'Actual T_{end} (s)')
plt.ylabel(r'Predicted T_{end} (s)')
plt.axis('equal')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
````

##  Ajustement comportemental : comparaison des prédictions brutes et ajustées

Cette cellule applique la dernière étape du pipeline comportemental : l’ajustement basé sur la variance résiduelle, permettant d’estimer le **seuil comportemental minimal robuste**
$$T_{end}^{\text{adj}}$$
indépendant du bruit perceptif individuel.

L’objectif est de comparer :

* la prédiction brute du modèle analytique (orange),
* la prédiction ajustée (violet),
* la ligne idéale (y=x) (rouge pointillée).


###  1. Calcul de l’ajustement comportemental

Pour chaque observation, la prédiction brute est corrigée en fonction de la météo :

$
T_{end}^{\text{adj}} = T_{end}^{\text{pred}} - 2\sigma_w + \mu_w
$

où :

* ($\sigma_w$) = écart-type des erreurs pour la météo (w)
* ($\mu_w$) = moyenne des erreurs pour la météo (w)

Cet ajustement :

* réduit systématiquement les valeurs pour intégrer une **marge de sécurité comportementale**,
* stabilise le seuil final contre les sur-estimations perceptives.

###  2. Binned trend : analyse de la forme globale

Comme dans la figure précédente, les données sont regroupées en 5 bins selon
$
T_{end}^{\text{true}}
$

Pour chaque bin, on trace :

* **orange** : prédiction brute du modèle analytique
* **violet** : prédiction ajustée comportementalement
* barres d’erreur = erreur standard dans le bin

###  3. Interprétation

Le graphique permet d’observer trois comportements distincts :

#### **1. Ligne orange (prédiction brute)**

* suit bien la diagonale,
* reflète la structure analytique du modèle,
* ne contient pas encore la correction comportementale.

#### **2. Ligne violette (prédiction ajustée)**

* est systématiquement en dessous de la prédiction brute,
* reflète la sécurité comportementale observée en VR,
* représente le **vrai seuil décisionnel** utilisé dans ton modèle final (celui fourni aux partenaires industriels).

####  **3. Points bleus**

* distribution brute du test set,
* montrent le bruit perceptif et les variations inter-participants.



Cette figure illustre clairement que le modèle final ne cherche pas à reproduire le comportement perceptif instantané, mais à prédire un **seuil robuste et conservateur** utilisé dans :

$
\text{Décision de traversée : } \text{Traverse si } TTC_{\text{real}} \ge T_{end}^{\text{adj}}
$


````python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd

# Assume df_bin contient 'y_true', 'y_pred' et 'weather'
stds = {
    'clear': 1.0,
    'night': 0.94,
    'rain': 0.72
}
mes = {
    'clear': 0.04,
    'night': 0.00,
    'rain': 0.00
}

# Calcul adjusted_time
df_bin['adjusted_pred'] = df_bin.apply(
    lambda row: row['y_pred'] - 2 * stds[row['weather']] + mes[row['weather']],
    axis=1
)

# Binning
nb_bins = 5
df_bin['bin'] = pd.cut(df_bin['y_true'], bins=nb_bins)

# Statistiques binned pour y_pred
bin_stats = df_bin.groupby('bin').agg(
    mean_y_true=('y_true', 'mean'),
    mean_y_pred=('y_pred', 'mean'),
    std_y_pred=('y_pred', 'std'),
    count=('y_pred', 'count')
).reset_index()
bin_stats['std_error'] = bin_stats['std_y_pred'] / np.sqrt(bin_stats['count'])

# Statistiques binned pour adjusted_pred
bin_stats_adj = df_bin.groupby('bin').agg(
    mean_y_true=('y_true', 'mean'),
    mean_adjusted_pred=('adjusted_pred', 'mean'),
    std_adjusted_pred=('adjusted_pred', 'std'),
    count=('adjusted_pred', 'count')
).reset_index()
bin_stats_adj['std_error'] = bin_stats_adj['std_adjusted_pred'] / np.sqrt(bin_stats_adj['count'])

# Plot
min_val = min(df_bin['y_true'].min(), df_bin['y_pred'].min(), df_bin['adjusted_pred'].min())
max_val = max(df_bin['y_true'].max(), df_bin['y_pred'].max(), df_bin['adjusted_pred'].max())

plt.figure(figsize=(6,6))
palette = sns.color_palette("deep")

scatter_test = palette[0]
binned_color = '#4c72B0'
adjusted_color = palette[4]

# --- TRAIN (raw predictions, no adjustment) ---
plt.scatter(
    df_train['y_true'], df_train['y_pred'],
    alpha=0.3,
    color='red',
    label='Train set'
)

# --- TEST scatter ---
plt.scatter(
    df_bin['y_true'], df_bin['y_pred'],
    alpha=0.9,
    color=scatter_test,
    label='Test set'
)


# Ligne y = x
plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=1.5, label=r'$y = x$')

# Courbe binned originale (orange)
plt.errorbar(
    bin_stats['mean_y_true'],
    bin_stats['mean_y_pred'],
    yerr=bin_stats['std_error'],
    fmt='o-', color=binned_color, ecolor=binned_color,
    elinewidth=2, capsize=4,
    label='Model Prediction'
)

# Courbe binned ajustée (violet)
plt.errorbar(
    bin_stats_adj['mean_y_true'],
    bin_stats_adj['mean_adjusted_pred'],
    yerr=bin_stats_adj['std_error'],
    fmt='s--', color=adjusted_color, ecolor=adjusted_color,
    elinewidth=2, capsize=4,
    label='Adjusted Prediction'
)

# Layout
plt.title(r'Predicted vs Actual Values')
plt.xlabel(r'Actual T_{end} (s)')
plt.ylabel(r'Predicted T_{end} (s)')
plt.axis('equal')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
````

Cette cellule examine **où et pourquoi** le modèle prédit moins précisément le seuil comportemental minimal
$$T_{end}$$
en analysant l’erreur absolue selon trois dimensions comportementalement pertinentes :

1. **la hauteur des participants**,
2. **la vitesse du véhicule **,
3. **la météo**.

L’objectif est d’identifier d’éventuels sous-groupes où les erreurs sont plus élevées, ce qui permet d’évaluer la robustesse du modèle et de comprendre la structure comportementale sous-jacente.


##  Influence de la hauteur des participants

Les participants sont regroupés en 5 catégories :
**157–167**, **167–171**, **171–176**, **176–180**, **180–190 cm**.

Le boxplot montre :

* une légère augmentation de l’erreur autour de 170–176 cm,
* une erreur plus faible pour les participants les plus grands (180–190 cm),
* un écart-type plus important dans les groupes intermédiaires, suggérant une plus grande variabilité comportementale.

 **Interprétation** :
La hauteur est une variable clé du modèle analytique. Les variations d’erreurs indiquent une sensibilité accrue dans les hauteurs intermédiaires, cohérente avec les différences perceptivo-motrices observées dans l’expérience VR.



##  Influence de la vitesse du véhicule

Les vitesses analysées sont : **20, 30, 40, 50, 60, 70 km/h**.

Observations principales :

* la vitesse **30 km/h** présente les plus fortes erreurs (et un écart-type très élevé),
* les vitesses élevées (60–70 km/h) montrent une erreur plus faible et moins dispersée,
* les vitesses intermédiaires (40–50 km/h) ont un comportement plus stable.

 **Interprétation** :
À 30 km/h, les situations ambiguës (ni trop rapides ni trop lentes) génèrent davantage d’incertitude comportementale, ce qui se reflète dans les données VR et dans le modèle.


##  Influence de la météo

Comparaison entre conditions : **clear**, **rain**, **night**.

Résultats :

* les erreurs sont légèrement plus faibles en **rain**,
* les conditions **night** et **clear** présentent des erreurs moyennes proches,
* la condition **clear** contient quelques valeurs extrêmes (max > 2.5).

 **Interprétation** :
La météo agit comme un **modulateur perceptif**, mais sa correction (via les (\alpha_w)) est bien capturée par le modèle.
Les différences d’erreurs entre conditions restent limitées, ce qui valide le choix d’un multiplicateur météo simple et analytique.
